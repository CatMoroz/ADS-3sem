```python
def bridges(n: int, g: list[list[int]]):
    color = [0]*n
    tup = [0]*n
    tin = [0]*n
    t=0
    ap = []
    cnt = 0
    def dfs(v : int, pv: int) -> None:
        nonlocal t
        nonlocal cnt
        color[v] = 1
        t+=1
        tin[v] = t
        tup[v] = tin[v]
        for u in g[v]:
            if pv == u:
                continue
            elif color[u] == 1: 
                tup[v] = min(tup[v], tin[u])
            elif color[u] == 0:
                dfs(u, v)
                cnt+=1
                if pv != -1 and tin[v] <= tup[u]:
                    ap.append(v)
        if pv != -1 and cnt >= 2:
            ap.append(v)
        color[v] = 2
    for v in range(n):
        if color[v] == 0:
            dfs(v, -1)
    return ap
```

```python
def list_edge_to_list_sm(list_edge:list[tuple[int, int]], n: int) -> list[list[int]]:
    list_sm = [[] for _ in range(n)]
    for edge in list_edge:
        list_sm[edge[0]].append(edge[1])
    return list_sm
```

```python
def list_sm_to_list_edge(list_sm:list[list[int]], n: int) -> list[tuple[int, int]]:
    list_edge = []
    for i in range(n):
        for j in list_sm[i]:
            list_edge.append((i, j))
    return list_edge
```

```python
def islands(n:int, list_sm: list[tuple[int, int]]):
    mosts = bridges(n, list_sm)
    edges = list_sm_to_list_edge(list_sm, n)

    edge_new = []
    for i in edges:
        if i in mosts or (i[-1], i[0]) in mosts:
            continue
        edge_new.append(i)

    list_sm_new = list_edge_to_list_sm(edge_new, n)

    color = [0]*n
    def dfs(v : int, i:int) -> None:
        color[v] = i
        for u in list_sm_new[v]:
            if color[u] == 0: 
                dfs(u, i)
    for i in range(n):
        if color[i] == 0:
            dfs(i, i+1)
    
    res = []
    for i in set(color):
        sw = []
        for j in range(n):
            if color[j] == i:
                sw.append(j)
        res.append(sw)
    return res

            
```
